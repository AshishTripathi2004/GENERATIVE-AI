```python
#variables
#containers that hold values 

age = 10
name = "Ashish"
std = 5

print(age, name, std)
```

```python
##Naming Convention
##variable names should be descriptive
##variable names can contain numbers, letters and underscore
##it can start with letter or underscore but never with a number
##it should not be a kwyword

age = 10

user_name = "Ashish"

_std = 20

print(age, user_name, _std)
```

```python
##Dynamic Typing
##Python allows to change the type of the variable as the program executes
##Python is dynamically typed language
## during the runtime the type is assigned

a = 10
res = True

print(type(a))
print(type(res))
```

```python
#Type conversion

a = 10
a_str = str(a)
print(type(a_str))
```

```python
##strings containing numerical values can be converted into int
age = "25"
age = int(age)
print(type(age))
```

```python
##int to float
##type promotion
a = 25
a = float(a)
print(type(a))

##float to int
##int is more restrictive than float
##leads to loss of data (beyond the decimal point)
a = 12.34
print(a, type(a))
a = int(a)
print(a, type(a))
```
