### Definition of Data Types  

- **Data types** classify data and define how it is used in a program.  
- They determine:  
  - The operations that can be performed on the data.  
  - The possible values the data can take.  
  - The amount of memory required to store the data.  


```python
##INTEGERS and FLOATING POINT
a = 12
print(type(a))
b = 12.1232
print(type(b))
print(type(a+b))
```

```python
##STRINGS
a = "ABCDEF"
print(type(a))
```

```python
##BOOLEAN VALUES -- TRUE OR FALSE
a = True
print(type(a))
a = 10
b = 20
print(type(a == b))
##empty string is treated as a falsy value
a = ""
print(bool(a))
```

```python
## STRING OPERATIONS
## string concatenation
result = "hello"
print(result + str(5))

## string multiplication
print(result*5)
```
