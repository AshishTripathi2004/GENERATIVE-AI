```python
## ARITHMATIC OPERATORS
a = 10
b = 5

print(a+b)
print(a-b)
print(a*b)
print(a/b)
print(a%b)#modulus; returns the remainder
print(a//b)#floor division; returns oly the integral part
print(a**b)#exponentiation; returns the power
```

```python
## COMPARISON OPERATORS
a = 10
b = 20

print( a == b)
print( a <= b)
print( a > b )
print( a != b )

## character by character matching
## ascii value based
str1 = "abcdef"
str2 = "Abcdef"
print(str1 == str2)
print(str1 > str2)
```

```python
## LOGICAL OPERATORS
a = True
b = False

print(a and b)
print(a or b)
print(not a)
```
