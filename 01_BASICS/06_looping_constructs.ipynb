```python
## for loops for iterating over a range
## range function
## range(start, end, step)
## start inlusive; end exclusive always
## default start = 0; step = 1

for i in range(5):
    print(i,end = " ")

print("\n")##new line character

for i in range(1,6):
    print(i, end = " ")
```

```python
#print all odd numbers upto n
n = int(input("Enter n:"))
for i in range(1,n+1,2):
    print(i, end=" ")
```

```python
## negative step size
for i in range(10, 0, -1):
    print(i, end=" ")
```

```python
## for loop for traversing a string
## string is an iterable
str = "Ashish"
for i in str:
    print(i)
```

```python
## while loops continues execution until a specific condition is true
count = 1
while(count<=5):
    print("Good Morning")
    count=count+1##incrementing is required for loop termination; else it will end up in an infinite loop
```

```python
## loop control statements
## break
## break -- stops the execution and jumps out of the loop to the next statement in the flow of execution

while(True):
    print("Hello welcome to this game...")
    cont = input("Do you wanna continue(y/n):").lower()
    if(cont == "n"):
        break
```

```python
##continue
##skips the execution for a particular condition
for i in range(1,11):
    if(i%2 == 0):## skips the iteration for all even numbers
        continue
    print(i, end=" ")
```

```python
##pass
##simply a placeholder for future logic
##useful for maintaining syntactic and logical structure of the code
for i in range(1,12):
    if(i%2==1):
        pass
    else:
        print(i, end = " ")
```

```python
## nested loops
## loop within a loop
## the depth of nesting could be to any level

for i in range(1,4):
    for j in range(1,4):
        print(f"i:{i} and j:{j}") ## formatted string literal; f-string
```

```python
#sum of first n natural numbers
n = int(input("Enter n:"))

sum=0
for i in range(1,n+1):
    sum+=i
print(sum)
```

```python
#printing prime numbers
#using seive of eratosthenes
lst = [1]*101

lst[0]=0
lst[1]=0

for i in range(0,101):
    if(lst[i]==1):
        j=i*i
        while(j<101):
            lst[j]=0
            j+=i

for i in range(0,101):
    if(lst[i]==1):
        print(i,end=" ")

    
```
