```python
## lists 
## ordered mutable collection of items
## can store multiple data types together
## elements of a list can be accessed via indices

lst=[1,2,3,4,5,True,"hello world",3.453,[1,2,3],(11,22,33)]
print(type(lst))
print(lst)
```

```python
## accessing list items
## done with the help of indices
fruits = ["apple", "banana", "cherry", "guava", "mango", "watermelon"]

for i in fruits:
    print(i)

for i in range(0,len(fruits)):
    print(fruits[i])
```

```python
## list slicing
## the last index is always excluded
print(fruits[1::])
print(fruits[::])
print(fruits[2:3])
print(fruits[::-1])
print(fruits)
```

```python
## negative indexing
## last index of the list == -1 and it keeps on decreasing to -len(lst)
print(fruits[-1])
print(fruits[-1:-7:-1])
print(fruits[::-2])
print(fruits[::-3])
print(fruits[::-4])
print(fruits[::-5])
print(fruits[::-6])
print(fruits[::-11])
```

```python
## lists are mutable in nature
## hence they could be modified

fruits = ["apple", "banana", "cherry"]
print(fruits)
fruits[2] = "mango"
print(fruits)
```

```python
lst = ["a"]
lst[1:]="bcdefghijklmnopqrstuvwxyz"
print(lst)
```

```python
## list methods
cities = ["Tokyo", "New Delhi", "Atlanta", "Abuja", "Oslo"]
print(cities)
cities.append("Los Angeles")## inserts the element at the end of the list
print(cities)
```

```python
##insert an element at the desired index
##use insert method
##shifts the remaning elements
cities.insert(1, "Mumbai")
print(cities)
```

```python
## removes the first occurence of the specified string
cities.remove("Mumbai")
print(cities)
```

```python
## pop removes the element at last index and returns it
ele = cities.pop()
print(ele)
print(cities)
```

```python
## get the index of the first occurence of the element
## if not present it throws an error
print(cities.index("New Delhi"))
```

```python
## count the occurrences of a particular element in the list
print(cities.count("Mumbai"))
```

```python
## sorting as per the ASCII value of the characters if strings are present
## can only sort similar or related data types like string and bool cannot be compared
cities.sort()
print(cities)
```

```python
## reversing a list
cities.reverse()## in-place reversal
print(cities)
cities = cities[::-1]## reverses by creating a copy
print(cities)
```

```python
## iterating over a list 
for city in cities:
    print(city)
```

```python
## iterating with indices
for [index,city] in enumerate(cities):
    print(index,city)
```

```python
## list comprehension
## basic syntax [result item in iterable if condition]

lst = [x**2 for x in range(1,11)]
print(lst)

lst = [x**2 for x in range(1,11) if x%2 == 0]
print(lst)
```

```python
##nested list comprehension
a = [1,2,3,4,5]
b = [11,22,33,44,55]
sum = [x+y for x in a for y in b]
print(sum)
```

```python
## add only elements at the corresponding indices
a = [1,2,3,4,5]
b = [11,22,33,44,55]
sum = [x+y for i1,x in enumerate(a) for i2,y in enumerate(b) if i1 == i2]
print(sum)
```
