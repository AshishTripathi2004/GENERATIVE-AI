```python
## tuples are similar to lists except they are immutable
tup = (1,2,3,4)
print(type(tup))
```

```python
## converting list to tuples
lst = [1,2,3,4,5]
tup = tuple(lst)
print(tup)

## converting tuple to list
nlst = list(tup)
print(nlst)
```

```python
## tuple operations
tup = (1,2,3,4,5)
tup1 = (6,7,8,9,10)

print(tup + tup1)
```

```python
tup*3 ##concatenates the same tuple for three times === (tup + tup + tup)
```

```python
## immutable nature of tuples; tuples do not support item assignment
## tup[1]=22

tup = list(tup)
tup[1]=22
tup = tuple(tup)
print(tup)
```

```python
## packing tuples
new_tup = 1,2,3,3.14,4,5
print(new_tup)
```

```python
## unpacking tuples
tup = ("1","2","3")
a,b,c = tup
print(a,b,c)
```

```python
##unpacking tuples with star notation
first,*middle,last = new_tup
print(first)
print(middle)##packs the remaining unpacked tuple elements into a list
print(last)
```

```python
##nested tuple
tup = ((1,2,3),("a","b","c"),(True, False, True))
print(tup)
for sub in tup:
    for i in sub:
        print(i, end=" ")
    print()
```
