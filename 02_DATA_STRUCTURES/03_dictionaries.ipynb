```python
## dictionaries
## store key-value pairs

## declaring a decitionary

new_dict = {}
print(type(new_dict))
new_dict = dict()
print(type(new_dict))
```

```python
## keys should be unique ; values could be duplicate
## otherwise it will over-write the data with the most recent value leading to loss of data
student = {"name":"Ashish",
           "age":21,
           "grade":"A"}
print(student)
```

```python
## accessing dictionary elements
## if key is not present it usually returns None
print(student["name"])
print(student["grade"])

## we can also pass a default value in case if key doesnt exists
print(student.get("grade"))
print(student.get("age"))
print(student.get("last_name","Not Availaible"))## returns not availaible
```

```python
## dictionaries are mutable data types
## allow updation deletion modification addition 
student["name"]="AASSHHIISSHH"
print(student)

student["last-name"]="Tripathi"
print(student)

## deleting a key-value pair using the del keyword
del student["last-name"]
print(student)
```

```python
keys = student.keys()
print(keys)## list of keys
values = student.values()
print(values)## list of values associated with keys
items = student.items()
print(items)## list of tuples of key-value pairs
```

```python
## shallow copy
student_cpy=student
print(student)
print(student_cpy)

## shallow copy means they reference to the same memory locations
## hence any changes will be reflected in both the dictionaries
student["name"]="Ashish"
print(student)
print(student_cpy)
```

```python
## solution is to create a deep copy
## an independent dictionary altogether to ensure isolation

student_cpy=student.copy()
print(student)
print(student_cpy)

student["name"]="AA"
print(student)
print(student_cpy)

## in case if nesting is present use copy.deepcopy() for creating a deep copy
```

```python
## iterating over keys
for keys in student.keys():
    print(student[keys])
```

```python
for values in student.values():
    print(values)
```

```python
for keys,values in student.items():
    print(f"{keys}:{values}")
```

```python
## nested dictionaries
students={
    "student1":{"name":"AAA","age":23},
    "student2":{"name":"BBB","age":34}
}

print(students["student1"]["age"])
print(students["student2"]["name"])
```

```python
##iterating over nested dictionaries
for keys,values in students.items():
    print(f"{keys}:{values}")
    for key,value in values.items():
        print(f"{key}:{value}")
```

```python
## dictionary comprehension
squares={x:x**2 for x in range(1,6)}
print(squares)
```

```python
## square of only even numbers
cubes_even={x:x**3 for x in range(1,6) if x%2==0}
print(cubes_even)
```

```python
## lets add both if and else
dict1 = {x:x**2 if x%2==0 else x**3 for x in range(1,11)}
print(dict1)
```

```python
## frequency map
nums = [1,1,2,2,3,3,4,4,4,4,5,5,6]
frequency = {}

for i in nums:
    if i in frequency:
        frequency[i]+=1
    else:
        frequency[i]=1

print(frequency)
```

```python
## merge two dictionaries
dict1 = {"a":1,"b":2}
dict2 = {"c":3,"d":4} 
merged = {**dict1, **dict2}## use the unpacking operator
## overides common keys by preferring the last unpacked dictionary
print(merged)
```
