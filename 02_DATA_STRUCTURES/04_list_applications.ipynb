```python
## creating a to-do list for managing tasks
todo = ["Clean the house", "Make dinner", "Pay Water Bills"]

## add more tasks
todo.append("Do the Laundry")
todo.append("Order Groceries")

##remove a task that is finished
todo.remove("Make dinner")

##set up alerts or reminders
if("Pay Water Bills" in todo):
    print("Do not forget to pay the hydro bills...")

##procure a list of still to do tasks
print("Tasks that still remain:")
for task in todo:
    print(f"--{task}")
```

```python
## organising student grades
grades = [89,99,100,34,56,78,88,76]

## adding grades
grades.append(90)

## updating the grades of the respective roll_no
grades[2]=98

##preparing a report for performance of class and indivduals
maxi = max(grades)
mini = min(grades)
avg = round(sum(grades)/len(grades),2)

print("CLASS PERFORMANCE")
print(f"Highest Marks:{maxi}")
print(f"Lowest Marks:{mini}")
print(f"Average Marks:{avg}")
```

```python
## inventory management
inventory = ["mango", "apple", "pineapple","cherry","orange"]

##adding a new product
inventory.append("strawberry")
inventory.append("blueberry")

##removing items that went out of stock
inventory.remove("orange")

##check if an item is in-stock or out-of-stock
if "orange" in inventory:
    print("Yes....present...")
else:
    print("Oops....Out of stock...")

##getting the inventory
print("Items in inventory...")
for item in inventory:
    print(item)
```
