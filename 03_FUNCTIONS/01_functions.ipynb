```python
##create a function to check whether a number is odd or even
def checkOddEven(num):
    """This function checks whether a number is odd or even"""
    if num%2==0:
        return "EVEN"
    return "ODD"

## calling the function
print(checkOddEven(4))
print(checkOddEven(67))
```

```python
## function with multiple parameters
def add(a,b):
    return a+b

print(add(23,34))
print(add(-1,8))
```

```python
## default parameters
def greet(name="Learner"):
    print(f"Hello {name}...Welcome to the world of functions...")

greet("Ashish")
greet()
```

```python
## variable length arguments
## positional arguments

def print_items(*args):
    for i in args:
        print(i)

print_items(1,2,2,3,4,5,6.12)
```

```python
## keyword arguments
## parameters will be in the form of key-value pairs
def print_details(**kwargs):
    for key,value in kwargs.items():
        print(f"{key}:{value}")

print_details(name="Ashish", std=10, stream="PCM", age=15)
```

```python
## we can use both positional and keyword arguments together
## positional arguments cannot come after keywords arguments
def print_all(*args, **kwargs):
    for i in args:
        print(f"Positional arguments:{i}")
    for key,value in kwargs.items():
        print(f"Kwywords arguments:[{key}:{value}]")

print_all(1,2,3,4,5,name="Ashish", std=10, stream="PCM", age=15)
```

```python
## return multiple params
def sum_product(a,b):
    return a+b, a*b ##packs it in a tuple by default

print(sum_product(8,9))
print(type(sum_product(1,2)))
```
