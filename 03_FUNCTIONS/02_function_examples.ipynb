```python
## temperature conversion
def tempConvert(temp,unit):
    ##celsius to fahrenheit
    if unit=="C":
        return temp*(9/5)+32
    elif unit=="F":##fahrenheit to celsius
        return (temp-32)*(5/9)
    else:
        return None
    
    

print(tempConvert(25,"C"))
print(tempConvert(77,"F"))
```

```python
## check if a password is strong or not
## strong password has the following characteristics
## length must be greater than 8
## must contain digits
## must contain letters
## must both uppercase and lowercase
## must contain special characters

def checkPassword(password):
    if len(password)<8:
        return False
    if not any(char.isdigit() for char in password):
        return False
    if not any(char.islower() for char in password):
        return False
    if not any(char.isupper() for char in password):
        return False
    if not any(char in "@#!&*()_$%^-+=?" for char in password):
        return False
    return True

checkPassword("Ashish@1989")
```

```python
## total cost of items in a shopping cart and the receipt
def shoppingCart(cart):
    payableAmount = 0
    print(f"{'ITEM':<10}{'PRICE':<10}{'QUANTITY':<10}{'AMOUNT':<10}")
    print("=" * 40)  # Separator line

    for product in cart:
        tot_amount = product["price"] * product["quantity"]
        payableAmount += tot_amount
        print(f"{product['name']:<10}{product['price']:<10}{product['quantity']:<10}{tot_amount:<10}")

    print("=" * 40)
    print(f"{'Total payable amount:':<30}{payableAmount:<10}")



##cart data
##list of dictionaries
cart=[
    {"name":"Apple","price":20,"quantity":5},
    {"name":"Coffee","price":200,"quantity":2},
    {"name":"Flour","price":350,"quantity":1},
    {"name":"Water","price":30,"quantity":10}
]

shoppingCart(cart)
```

```python
##check palindrome
##case insensitive
def checkPalindrome(word):
    word = word.lower().replace(" ","")
    return word == word[::-1]

print(checkPalindrome("Malay  alam"))
print(checkPalindrome("abcded"))
```

```python
##factorial using recursion
def factorial(n):
    if(n==0):
        return 1
    return factorial(n-1)*n

print(factorial(5))
```

```python
##count the frequency of words from the contents of a file
def count_word_frequency(file_path):
    word_count=dict()
    with open(file_path,'r') as file:
        for line in file:
            words=line.split()
            for word in words:
                word = word.lower().strip("!.,?:;'\" ")
                if(len(word)):
                    word_count[word]=word_count.get(word,0)+1
    return word_count

filepath = "./sample.txt"
word_frequency = count_word_frequency(filepath)
print(word_frequency)
```
