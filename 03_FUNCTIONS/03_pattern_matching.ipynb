```python
import re
```

```python
string = "I love python"
pattern = r"python"

match = re.search(pattern,string)## matches for the presence of the pattern anywhere in the string
if match:
    print("Pattern found")
else:
    print("Pattern not found")
```

```python
##validate the phone number 
##+91-XXXXXXXXXX is the correct format
def checkPhoneNumber(phone):
    pattern =r'^\+91[-]?[0-9]{10}'
    if re.fullmatch(pattern,phone):
        return True
    return False

checkPhoneNumber("+912321038607")
```

```python
##validate email address
##abcdef@example.com

def validateEmailAddress(email):
    pattern = r'^[a-z0-9]+@[a-z]+\.[a-z]+(\.[a-z]+)*$'
    if re.fullmatch(pattern,email):
        return True
    return False

validateEmailAddress("ashishtripathi0306")
```
