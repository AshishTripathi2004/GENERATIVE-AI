```python
##anonymous functions
##accepts using any number of arguments but allows only one expression 
even = lambda num:num%2==0
print(even(34))

square = lambda x:x**2
print(square(7))
```

```python
add = lambda x,y,z:x+y+z
print(add(1,2,3))
print(add(11,22,33))
```

```python
##maximum of two numbers
maxi = lambda x,y:x if x>y else y
print(maxi(11,22))
print(maxi(22,1))
```

```python
##check sign
check = lambda x: "Positive" if x>0 else ("Negative" if x<0 else "Zero")
print(check(1000))
print(check(-100))
print(check(0))
```
