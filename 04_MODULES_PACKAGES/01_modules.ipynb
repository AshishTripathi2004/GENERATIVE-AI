```python
##compute area of a sphere
def getArea(radius):
    area = round(4*math.pi*pow(radius,2),4)
    print(area,"sq units")

getArea(4)
```

```python
import random
```

```python
##prints a random float between 0 and 1(exclusive)
print(random.random())
##prints a random int between a and b(inclusive)
print(random.randint(1,10))
##printa a random float between a and b
print(random.uniform(1,10))
##step value can be specified
print(random.randrange(2,10,2))
```

```python
##shuffle and choice
cards = ["ace","spade","heart","club"]
print(random.choice(cards))
random.shuffle(cards)
print(cards)
```

```python
##seeding randomness
##seed value = 42 as a nod to Hitchhiker’s Guide
random.seed(42)
print(random.randint(1,100))
print(random.randint(1,100))
print(random.randint(1,100))
print(random.randint(1,100))
```
