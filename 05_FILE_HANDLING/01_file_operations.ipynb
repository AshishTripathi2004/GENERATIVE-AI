```python
## reading a text file all at once
with open("example.txt","r") as file:
    content=file.read()
    print(content)
```

```python
##reading a file line by line
with open("example.txt","r") as file:
    for line in file:
        print(line.strip())##removes the new line character
```

```python
##write into the file
with open("example.txt","w") as file:
    file.write("This is the new content.\n")
    file.write("The contents have been overriden.\n")
with open("example.txt","r") as file:
    for line in file:
        print(line.strip())
```

```python
##writing a file without overriding
##use append operation
with open("example.txt","a") as file:
    file.write("This is the appended line...\n")
```

```python
##writing a list of lines
shoppingList = ["coffee","tea","sugar","flour","eggs","wafers","oil"]
with open("example.txt","a") as file:
    file.write("SHOPPING LIST\n")
    file.write("="*40)
    file.write("\n")
    for item in shoppingList:
        file.write(f"--{item}\n")

## use writelines
with open("example.txt","a") as file:
    file.writelines(shoppingList)
```

```python
##reading and writing binary files
content = b"\x00\x01\x02"
with open("example.bin","wb") as file:
    file.write(content)
```

```python
with open("example.bin","rb") as file:
    content = file.read()
    print(content)
```

```python
## reading the data from a source file
## copying the data into a destination file
with open("source.txt","r") as file_read:
    contents = file_read.read()
    ##if file doesnot exists it creates a destination file
    with open("destination.txt","w") as file_write:
        file_write.write(contents)
```

```python
##read a text file and count the number of lines
lines=0
with open("source.txt","r") as file:
    contents = file.readlines()
    for line in contents:
        lines+=1
        print(f"LINE-{lines} : {line.strip()}")

print(f"Total number of lines are {lines}")
```

```python
##count all words in the file
words = 0
with open("source.txt","r") as file:
    while True:
        line = file.readline()
        if not line:
            break
        words += len(line.strip().split(" "))
print(words)
```

```python
##count all characters
##assume correct format
chars = 0
with open("source.txt","r") as file:
    while True:
        line = file.readline()
        if not line:
            break
        chars += sum([len(word) for word in line.strip().split(" ")])
print(chars)
```
