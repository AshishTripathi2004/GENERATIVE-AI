
#### What are Exceptions?  
Exceptions are **events that disrupt the normal flow of execution** of a program.  

#### When Are Exceptions Raised?  
Exceptions are **raised when errors are encountered** during program execution.  


```python
## try and except block 
try:
    a=b
except NameError as ex:##NameError is a class of error that defines the error that has occurred
    print(ex)
```

```python
try:
    a=int(input("a="))
    b=int(input("b="))
    print(a/b)
except ZeroDivisionError as er:
    print(er)
    print("Enter a denominator other than 0")
```

```python
##main exception class
##Exception superclass-->base class of exceptions

try:
    a=int(input("a="))
    b=int(input("b="))
    print(a/b)
except ZeroDivisionError as er:
    print(er)
    print("Enter a denominator other than 0")
except Exception as er:
    print(er)
    print("Exception caught")
```

```python
## try except and else block
## try--contains code which might contain an exception
## except--containes code that runs is exception is raised
## else--containes code that runs if no exception occurs
## finally--always executes when exception happens or not
## useful for database handling

try:
    a=int(input("a="))
    b=int(input("b="))
    result = a/b
except ZeroDivisionError as er:
    print(er)
    print("Enter a denominator other than 0")
except ValueError as er:
    print(er)
    print("Invalid number")
else:
    print(result)
finally:
    print("everything done...")
```
