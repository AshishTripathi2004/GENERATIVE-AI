```python
## Inheritance
## Parent class
class Car:
    
    def __init__(self,type,engineType,capacity):
        self.type = type
        self.engineType = engineType
        self.capacity = capacity

    def getType(self):
        print(f"This car is a {self.type}")

## Tesla inherits the property of the class Car
class Tesla(Car):

    def __init__(self,type,engineType,capacity,brand,selfDriving):
        super().__init__(type,engineType,capacity)
        self.brand = brand
        self.selfDriving = selfDriving

    def isSelfDriving(self):
        if(self.selfDriving):
            print("Yes..self-driving")
        else:
            print("No...not self-driving")


car1 = Tesla("sedan","electric",5,"Tesla",True)
car1.getType()
car1.isSelfDriving()

car2 = Tesla("wagon","electric",4,"Tesla",False)
car2.getType()
car2.isSelfDriving()


```
