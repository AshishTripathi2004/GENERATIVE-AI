```python
## Method overriding
## allows the child class to override a function already defined in the parent class

class Animal:
    def speak(self):
        return "Animal-sound"
    
class Dog(Animal):
    def speak(self):
        return "woof"

class Cat(Animal):
    def speak(self):
        return "meow"
    
dog = Dog()
print(dog.speak())

cat = Cat()
print(cat.speak())


##function demonstrating polymorphism
def animal_speak(animal):
    print(animal.speak())

animal_speak(dog)
animal_speak(cat)
```

```python
### polymorphism with methods
class Shape:
    def area(self):
        """This function computes area"""
        pass

class Rectangle(Shape):
    def __init__(self,height,width):
        self.height = height
        self.width = width
    
    def area(self):
        print(f"The area of the rectangle is: {self.height*self.width} sq. units")

class Circle(Shape):
    def __init__(self,radius):
        self.radius = radius

    def area(self):
        print(f"The area of the circle is: {round(math.pi*math.pow(self.radius,2),2)} sq. units")

class Square(Rectangle):
    def __init__(self,side):
        Rectangle.__init__(self,side,side)

rec1 = Rectangle(12,25)
rec1.area()

cir1 = Circle(5)
cir1.area()

sq1 = Square(5)
sq1.area()
```

```python
## Abstract base class
## cannot be instantiated
## child must implement the functions of abstract base class
## enforces structure in child classes
from abc import ABC, abstractmethod

class Vehicle(ABC):
    @abstractmethod
    def engine_start(self):
        """To be overidden in child classes"""
        pass

class Car(Vehicle):
    def engine_start(self):
        print("Car engine started....ghrrrrr")

class Motorcycle(Vehicle):
    def engine_start(self):
        print("Motorcycle engine started....vrooooom")

mtr = Motorcycle()
car = Car()

mtr.engine_start()
car.engine_start()
    
```
