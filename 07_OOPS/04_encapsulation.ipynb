```python
## access modifiers 
## private,public,protected

class Person:
    def __init__(self,name,age,gender,rank):
        self.__name=name     ##private variables -- cannot be accessed outside the class
        self.__age=age       ##private variables 
        self.gender=gender   ##public variables  --can be freely accessed
        self._rank=rank      ##protected variables -- cannot be accessed outside the class but can be inherited


obj = Person("Ashish",23,"Male",1)
print(dir(obj))

## private: attribute ---> _classname__attribute
## protected: attribute ---> _attribute
## public: attribute --->attribute

print(obj._Person__age)
print(obj._rank)
print(obj.gender)

        
```

```python
##getter and setter functions can be used to access and modify the private variables in a simple manner
## getter get_attribute; setter set_attribute

class Student:
    def __init__(self,name,age,marks):
        self.__name = name
        self.__age = age
        self.__marks = marks

    def get_name(self):
        return self.__name
    
    def set_name(self,name):
        self.__name = name

    def get_age(self):
        return self.__age
    
    def set_age(self,age):
        if(age<0):
            print("Age cannot be less than 0....")
            return
        self.__age = age

    def get_marks(self):
        return self.__marks
    
    def set_marks(self,marks):
        self.__marks = marks

stud = Student("Ashish",20,25)
print(stud.get_marks())
stud.set_marks(40)
print(stud.get_marks())
stud.set_age(-1)
```
