```python
## hiding the complex implementational details and displaying only relevant information
## helps in reducing programming complexity and effort
from abc import ABC, abstractmethod

class Vehicle(ABC):
    @abstractmethod
    def engine_start(self):
        """To be overidden in child classes"""
        pass

class Car(Vehicle):
    def engine_start(self):
        print("Car engine started....ghrrrrr")

class Motorcycle(Vehicle):
    def engine_start(self):
        print("Motorcycle engine started....vrooooom")

mtr = Motorcycle()
car = Car()

mtr.engine_start()
car.engine_start()
```
