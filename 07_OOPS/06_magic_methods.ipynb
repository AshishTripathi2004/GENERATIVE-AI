```python
## magic methods
## also known as dunder methods 
## double underscore


'''
 __init__  - returns a new instance of the class.
 __str__   - Defines the string representation for print() and str().
 __repr__  - Returns a developer-friendly string representation.
 __add__   - Defines behavior for addition (+).
 __sub__   - Defines behavior for subtraction (-).
 __mul__   - Defines behavior for multiplication (*).
 __truediv__ - Defines behavior for division (/).
 __floordiv__ - Defines behavior for floor division (//).
 __mod__   - Defines behavior for modulus (%).
 __pow__   - Defines behavior for exponentiation (**).

'''

class Person:
    def __init__(self,age):
        self.age = age

    def __repr__(self):
        return "This is from the repr method"
    
    def __str__(self):
        return "This is from the str method"

per = Person(23)
## calls the __str__method and prints a string representation
## it can also call the __repr__ method
print(per)
dir(per)
```

```python
## operator overloading
class Vector:
    def __init__(self,x,y):
        self.x = x
        self.y = y

    def __add__(self,other):
        return Vector(self.x + other.x, self.y+other.y)
    def __sub__(self,other):
        return Vector(self.x - other.x, self.y-other.y)
    def __mul__(self,other):
        return Vector(self.x*other.x, self.y*other.y)
    def __eq__(self,other):
        return (self.x == other.x) and (self.y == other.y)
    def __pow__(self, other):
        return self.x * other.y - self.y * other.x 
    def __repr__(self):
        return f"Vector:{self.x} {self.y}"
    
A = Vector(9,2)
B = Vector(4,-1)

C = A + B
print(C)

C = A - B
print(C)

C = A * B
print(C)

print(A == B)
print(A ** B)
```
