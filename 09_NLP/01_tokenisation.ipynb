```python
from nltk.tokenize import sent_tokenize
```

```python
corpus = """The Himalayas are the tallest mountain range in the world.
These mountains have an astonishing average height of 1200 meters.
These mountains are the crown of India.
"""
```

```python
## sentences are recognised using punctuation marks like . or !
sentences = sent_tokenize(corpus)
print(sentences)
for sentence in sentences:
    print(sentence)
```

```python
from nltk.tokenize import word_tokenize
```

```python
## every word, punctuation mark and number is treated as a word itself
words = word_tokenize(corpus)
print(words)
```

```python
for sentence in sentences:
    print(word_tokenize(sentence))
```

```python
from nltk.tokenize import TreebankWordTokenizer
```

```python
## full stop is treated as a part of the last word
## except for the last word in the corpus

## create a callable object
tree = TreebankWordTokenizer()
for sentence in sentences:
    print(tree.tokenize(sentence))## use this to handle all the functions associated
```
