```python
## nltk.stem is used for stemming
## PorterStemmer is a simple technqiue to stem existing words
## still faces accuracy issues with certain words
from nltk.stem import PorterStemmer
```

```python
words = ["finally","finalised","sitting","finale","historic","historically","history","congratulations","loves","lover","loving"]

## create a callable object
stemming = PorterStemmer()
for word in words:
    print(f"{word}------->{stemming.stem(word)}")
```

```python
## another method is using regular expressions
## identify a list of words that need to be truncated in order to form the stem in the form of a regular expression

from nltk.stem import RegexpStemmer

## create a callable object
## identifies morphological affixes from the words
reg_stemming = RegexpStemmer("(?<!s)s$|ing$|able$|er$|en$|ed$", min=4)


## handles plural well
## (?<!s)s$
words = ["playing", "happily", "sitting", "tables", "reader", "golden", "coded", "at","dogs","cats","glass"]
for word in words:
     print(f"{word}------->{reg_stemming.stem(word)}")
```

```python
## snowball stemmer
## performs better than PorterStemmer
## offers multilanguage support

from nltk.stem import SnowballStemmer
```

```python
snowball = SnowballStemmer('english')

words = ["finally","finalised","stemming","sitting","finale","historic","historically","history","congratulations","loves","lover","lovingly","sportingly","sporting","lovingly"]
for word in words:
     print(f"{word}------->{snowball.stem(word)}")


## still inefficient to stem words as it makes them lose meanings
```

```python
## aggressive stemming using lancaster stemmer
## too harsh; should be avoided
from nltk.stem import LancasterStemmer

ls = LancasterStemmer()
for word in words:
     print(f"{word}------>{snowball.stem(word)}------->{ls.stem(word)}")
```
