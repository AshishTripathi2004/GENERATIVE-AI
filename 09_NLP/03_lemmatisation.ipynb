```python
## lemmatisation is used to convert the word into its root form
## performs better than stemming but is more time consuming 

from nltk.stem import WordNetLemmatizer
from nltk.stem import PorterStemmer
```

```python
lemmatizer = WordNetLemmatizer()
ps = PorterStemmer()
```

```python
'''
Parts of Speech (POS) Tags are also used;
Noun - 'n'
Verb - 'v'
Adjective - 'a'
Adverb - 'r'
'''

print(lemmatizer.lemmatize("going",pos='n'))
print(lemmatizer.lemmatize("going",pos='v'))
print(lemmatizer.lemmatize("congratulations",pos='v'))
```

```python
words = ["historically","history","eating","eats","eaten","wastes","wasted","ate","learnt","sang"]

for word in words:
   print(f"{word}------>{lemmatizer.lemmatize(word,pos='v')}")
```

```python
## why is lemmatisation slower
## WordNetLemmatizer is a thin wrapper around the WordNet corpus
## morphy() function to WordNet CorpusReader class to find words
import time


## measure stemming time
start = time.time()

for word in words:
   print(f"{word}------>{ps.stem(word)}")
print("STEMMING TIME TAKEN:",time.time()-start)

start = time.time()
for word in words:
   print(f"{word}------>{lemmatizer.lemmatize(word,pos='v')}")
print("LEMMATISATION TIME TAKEN:",time.time()-start)
```
