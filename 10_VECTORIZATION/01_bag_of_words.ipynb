```python
## read the csv file using pandas
## drop uneccessary columns and rename the columns as per convenience
messages = pd.read_csv("./spam.csv",encoding="ISO-8859-1")
messages = messages.drop(messages.columns[[2,3,4]],axis=1)
messages.columns = ["label","message"]
messages.head()
messages.shape
```

```python
import nltk
import numpy
import re
from nltk.stem import PorterStemmer
from nltk.corpus import stopwords
nltk.download("stopwords")
```

```python
corpus=[]
ps=PorterStemmer()
for i in range(0,len(messages)):
    ## re.sub( pattern, replacement, string)
    review = re.sub("^[a-zA-z]"," ",messages.iloc[i,1])
    review = review.lower()
    review = review.split()
    review = [ps.stem(word) for word in review if word not in stopwords.words('english')]
    review = " ".join(review)
    corpus.append(review)
```

```python
print(corpus)
```

```python
from sklearn.feature_extraction.text import CountVectorizer
```

```python
## implemeant bag of words using this CountVectorizer class
## max_features: select the most frequent features to be selected
## binary: false means normal Bow(default), true means binary bow
cv = CountVectorizer(max_features=100,binary=False,ngram_range=(2,3))
X = cv.fit_transform(corpus).toarray()
for row in X:
    print(row)
```
