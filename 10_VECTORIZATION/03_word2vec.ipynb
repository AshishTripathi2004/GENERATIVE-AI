```python
from gensim.models import Word2Vec, KeyedVectors
import gensim.downloader as api
```

```python
## pretrained model
## 3billion words; 300dimensions feature representation
## extremely heavy file nearly 1.6 GB
wv = api.load("word2vec-google-news-300")
```

```python
wv.most_similar("burger")
```
