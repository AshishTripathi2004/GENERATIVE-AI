```python
import tensorflow as tf
from tensorflow.keras.models import load_model
from tensorflow.keras.datasets import imdb
from tensorflow.keras.preprocessing import sequence
```

```python
model = load_model("simple_rnn_model.h5")
```

```python
model.summary()
```

```python
## use dictionary
word_index = imdb.get_word_index()

## pre-processing the review by using the dictionary
def preprocessText(review):
    review = review.lower()
    encoded_review = [word_index.get(word,2)+3 for word in review.split()]
    padded_review = sequence.pad_sequences([encoded_review],maxlen=500,padding='pre')
    return padded_review
```

```python
## prediction function
def prediction(review):
    preprocessed_input  = preprocessText(review)
    result = model.predict(preprocessed_input)
    print("Review:Positive review") if result[0][0]>0.5 else print("Review:Negative review")
    print(f"Sentiment score:{result[0][0]}")
```

```python
prediction("This movie was awesome and all the characters were amazing the story line was dynamic and worth watching")
```

```python
prediction("The movie was horrible and the characters were all boring and no plot twist and it is a flop")
```
