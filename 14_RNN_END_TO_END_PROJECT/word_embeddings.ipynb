```python
from tensorflow.keras.preprocessing.text import one_hot
```

```python
sentences = [
    "The sun sets behind the hills",
    "She smiled and walked away",
    "They danced under the moonlight",
    "I forgot my keys",
    "Time flies when you're having fun",
    "Please pass me the salt",
    "He ran faster than the wind"
]
```

```python
voc_size = 10000
one_hot_repr = [one_hot(word,voc_size) for word in sentences]
one_hot_repr
```

```python
## using embeddings
from tensorflow.keras.layers import Embedding, Input
from tensorflow.keras.preprocessing.sequence import pad_sequences
from tensorflow.keras.models import Sequential
```

```python
sent_len = 10
one_hot_repr = pad_sequences(one_hot_repr, maxlen=sent_len, padding='pre')
one_hot_repr
```

```python
dim = 10
model = Sequential()
model.add(Input(shape=(sent_len,)))
model.add(Embedding(input_dim=voc_size, output_dim=dim))
model.compile('adam', 'mse')
```

```python
model.summary()
```

```python
print(model.predict(one_hot_repr)[0][0])
```
