```python
import tensorflow as tf
from tensorflow.keras.models import load_model
from tensorflow.keras.preprocessing.text import Tokenizer
from tensorflow.keras.preprocessing.sequence import pad_sequences
import numpy as np
```

```python
## load the model
model = load_model("./LSTM_model.h5")
max_length = model.input_shape[1]

##read the hamlet 
with open("hamlet.txt","r") as file:
    text = file.read()

tokenizer = Tokenizer()
tokenizer.fit_on_texts([text])
```

```python
## pre-process the input
def predictNextWord(input_text):
    ## convert the text into lower case
    input_text = input_text.lower()
    ## initialise the tokenizer and tokenize into a sequence
    token_list = tokenizer.texts_to_sequences([input_text])[0]
    ## match the length to the length of the input
    if len(token_list)>max_length:
        token_list = token_list[-(max_length):]
    
    ## pad the sequences
    padded_input = pad_sequences([token_list], maxlen=max_length,padding="pre")
    prediction = model.predict(padded_input , verbose=0)
    predicted_word_index = np.argmax(prediction, axis=1)
    for word,index in tokenizer.word_index.items():
        if index == predicted_word_index:
            return word
    return None


```

```python
txt = "this is i haue"
print(predictNextWord(txt))
```
